```python
import random, math, itertools, statistics
from fractions import Fraction
from collections import Counter

from sympy import Rational
from sympy.stats import (Binomial, Poisson, Geometric, Uniform, Normal, Exponential,
                         density, P, E, variance)
import plotly.graph_objects as go

random.seed(42)

# Fungsi bantu kecil untuk merapikan dan menampilkan grafik Plotly
def atur(fig, judul, xlabel, ylabel):
    fig.update_layout(title=judul, xaxis_title=xlabel, yaxis_title=ylabel,
                      width=700, height=400, template="simple_white")
    fig.show()

print("random.randint(1,6)   :", random.randint(1, 6), "(lempar dadu)")
print("math.comb(5,2)        :", math.comb(5, 2), "(pilih 2 dari 5)")
print("itertools.product     :", list(itertools.product("AB", [1, 2])))
print("Fraction(6,36)        :", Fraction(6, 36), "(pecahan paling sederhana)")
print("Counter               :", Counter(["H", "T", "H", "H"]))
print("statistics.mean       :", statistics.mean([2, 4, 6]))
X = Binomial("X", 2, Rational(1, 2))
print("sympy P(X=1), 2 koin  :", P(X > 0) - P(X > 1), "(tepat 1 gambar dari 2 lemparan koin)")
```

```python
# 1) Manual (pecahan eksak)
manual = Fraction(6, 36)

# 2) Daftar semua kemungkinan
semua = list(itertools.product(range(1, 7), repeat=2))
cocok = [x for x in semua if sum(x) == 7]
hitung = len(cocok) / len(semua)

# 3) Simulasi 100.000 lemparan
n = 100_000
hasil = sum(1 for _ in range(n) if random.randint(1, 6) + random.randint(1, 6) == 7)
simulasi = hasil / n

print(f"Manual     : {manual} = {float(manual):.4f}")
print(f"Daftar     : {hitung:.4f}  ({len(cocok)} dari {len(semua)})")
print(f"Simulasi   : {simulasi:.4f}")
```

```python
tim = math.comb(10, 3)
print("a) Cara memilih tim :", tim)

jumlah_pin = 10 ** 4
print("b) Jumlah PIN       :", jumlah_pin)
print(f"   Peluang menebak  : {1/jumlah_pin:.4f} ({1/jumlah_pin*100:.2f}%)")

# Cek (a) dengan itertools
cek = len(list(itertools.combinations(range(10), 3)))
print("   Cek itertools (a):", cek)
```

```python
total, python, java, keduanya = 200, 120, 80, 50

p_python = Fraction(python, total)
p_keduanya = Fraction(keduanya, total)
p_java_given_python = p_keduanya / p_python
print(f"Manual   : P(Java|Python) = {p_java_given_python} = {float(p_java_given_python):.4f}")

# Simulasi: buat 200 mahasiswa, lalu pilih acak berulang
mhs = ["P&J"]*50 + ["P"]*70 + ["J"]*30 + ["-"]*50
sampel = random.choices(mhs, k=200_000)
bisa_python = [m for m in sampel if m in ("P&J", "P")]
simulasi = bisa_python.count("P&J") / len(bisa_python)
print(f"Simulasi : {simulasi:.4f}")
```

```python
p_spam = Fraction(3, 10)
p_g_spam = Fraction(8, 10)
p_g_normal = Fraction(1, 10)

p_gratis = p_g_spam * p_spam + p_g_normal * (1 - p_spam)
p_spam_given_g = (p_g_spam * p_spam) / p_gratis
print(f"Manual (Bayes) : {p_spam_given_g} = {float(p_spam_given_g):.4f}")

# Simulasi 300.000 email
n = 300_000
spam_dan_gratis = 0
gratis = 0
for _ in range(n):
    spam = random.random() < 0.30
    ada_gratis = random.random() < (0.80 if spam else 0.10)
    if ada_gratis:
        gratis += 1
        spam_dan_gratis += spam
print(f"Simulasi       : {spam_dan_gratis / gratis:.4f}")
```

```python
p = Fraction(1, 10)
minimal_satu = 1 - (1 - p) ** 3
ketiganya = p ** 3
print(f"Manual   : minimal 1 down = {float(minimal_satu):.4f} | ketiganya down = {float(ketiganya):.4f}")

n = 200_000
min1 = semua3 = 0
for _ in range(n):
    down = [random.random() < 0.1 for _ in range(3)]
    min1 += any(down)
    semua3 += all(down)
print(f"Simulasi : minimal 1 down = {min1/n:.4f} | ketiganya down = {semua3/n:.4f}")
```

```python
def peluang_sama(k):
    p_beda = 1.0
    for i in range(k):
        p_beda *= (365 - i) / 365
    return 1 - p_beda

print(f"Manual   : k=23 -> {peluang_sama(23):.4f}")

# Simulasi
n_uji = 20_000
hit = 0
for _ in range(n_uji):
    ulang_tahun = [random.randint(0, 364) for _ in range(23)]
    if len(set(ulang_tahun)) < 23:
        hit += 1
print(f"Simulasi : k=23 -> {hit/n_uji:.4f}")

# Grafik: peluang naik seiring jumlah orang
k = list(range(1, 71))
fig = go.Figure(go.Scatter(x=k, y=[peluang_sama(i) for i in k], mode="lines", line=dict(color="#d32f2f")))
fig.add_hline(y=0.5, line_dash="dash", line_color="gray")
fig.add_vline(x=23, line_dash="dash", line_color="gray")
atur(fig, "Paradoks Ulang Tahun", "Jumlah orang", "Peluang ada yang sama")
```

```python
n, p = 10, 0.1
X = Binomial("X", n, Rational(1, 10))

print(f"P(X=2)  = {float(density(X)(2)):.4f}")
print(f"P(X<=2) = {float(P(X <= 2)):.4f}")
print(f"E(X)    = {float(E(X)):.2f}  (rumus n*p = {n*p})")
print(f"Var(X)  = {float(variance(X)):.2f}  (rumus n*p*(1-p) = {n*p*(1-p):.2f})")

sampel = [sum(random.random() < p for _ in range(n)) for _ in range(100_000)]
print(f"Simulasi: rata-rata = {statistics.mean(sampel):.3f}, varians = {statistics.pvariance(sampel):.3f}")

x = list(range(0, n + 1))
frek = Counter(sampel)
fig = go.Figure()
fig.add_bar(x=x, y=[float(density(X)(i)) for i in x], name="Teori", marker_color="#d32f2f")
fig.add_scatter(x=x, y=[frek[i] / len(sampel) for i in x], mode="markers", name="Simulasi",
                marker=dict(color="black", size=9))
atur(fig, "Binomial(n=10, p=0.1)", "Jumlah paket hilang", "Peluang")
```

```python
lam = 4
X = Poisson("X", lam)

print(f"P(X=6)  = {float(density(X)(6)):.4f}")
print(f"P(X>8)  = {float(P(X > 8)):.4f}")
print(f"E(X)    = {float(E(X)):.2f}, Var(X) = {float(variance(X)):.2f}")

# Simulasi Poisson (algoritma Knuth): kalikan angka acak sampai nilainya <= e^(-lam)
def acak_poisson(lam):
    batas, k, hasil = math.exp(-lam), 0, 1.0
    while True:
        hasil *= random.random()
        if hasil <= batas:
            return k
        k += 1

sampel = [acak_poisson(lam) for _ in range(100_000)]
frek = Counter(sampel)
print(f"Simulasi: P(X=6) = {frek[6]/len(sampel):.4f}, P(X>8) = {sum(1 for s in sampel if s > 8)/len(sampel):.4f}")

x = list(range(0, 15))
fig = go.Figure(go.Bar(x=x, y=[float(density(X)(i)) for i in x], marker_color="#d32f2f"))
atur(fig, "Poisson(λ=4)", "Jumlah request per detik", "Peluang")
```

```python
p = 0.3
X = Geometric("X", Rational(3, 10))

print(f"P(X=3)  = {float(density(X)(3)):.4f}")
print(f"P(X<=3) = {float(P(X <= 3)):.4f}")
print(f"E(X)    = {float(E(X)):.3f}  (rumus 1/p = {1/p:.3f})")

def acak_geometrik(p):
    percobaan = 1
    while random.random() >= p:
        percobaan += 1
    return percobaan

sampel = [acak_geometrik(p) for _ in range(100_000)]
print(f"Simulasi: rata-rata percobaan = {statistics.mean(sampel):.3f}")

x = list(range(1, 16))
fig = go.Figure(go.Bar(x=x, y=[float(density(X)(i)) for i in x], marker_color="#d32f2f"))
atur(fig, "Geometrik(p=0.3)", "Percobaan ke-", "Peluang")
```

```python
a, b = 0, 10
X = Uniform("X", a, b)

print(f"P(X<3)  = {float(P(X < 3)):.4f}")
print(f"E(X)    = {float(E(X)):.2f}  (rumus (a+b)/2 = {(a+b)/2})")
print(f"Var(X)  = {float(variance(X)):.3f}  (rumus (b-a)^2/12 = {(b-a)**2/12:.3f})")

sampel = [random.uniform(a, b) for _ in range(100_000)]
print(f"Simulasi: P(X<3) = {sum(1 for s in sampel if s < 3)/len(sampel):.4f}")

x = [i / 10 for i in range(-20, 121)]
fig = go.Figure()
fig.add_histogram(x=sampel, histnorm="probability density", nbinsx=30, name="Simulasi",
                  marker_color="gray", opacity=0.5)
fig.add_scatter(x=x, y=[float(density(X)(i)) for i in x], mode="lines", name="Teori",
                line=dict(color="#d32f2f"))
atur(fig, "Uniform(0,10)", "Delay (detik)", "Kepadatan")
```

```python
mu, sigma = 200, 30
X = Normal("X", mu, sigma)

print(f"P(X>250)        = {float(P(X > 250)):.4f}")
print(f"P(170<X<230)    = {float(P(X < 230) - P(X < 170)):.4f}")
print(f"Batas 95% teratas = {statistics.NormalDist(mu, sigma).inv_cdf(0.95):.1f} ms")

sampel = [random.gauss(mu, sigma) for _ in range(100_000)]
n = len(sampel)
print(f"Simulasi: P(X>250) = {sum(1 for s in sampel if s > 250)/n:.4f}, "
      f"P(170<X<230) = {sum(1 for s in sampel if 170 < s < 230)/n:.4f}")

x = list(range(100, 301, 2))
fig = go.Figure()
fig.add_scatter(x=x, y=[float(density(X)(i)) for i in x], mode="lines", name="Kurva normal",
                line=dict(color="#d32f2f"))
x_ekor = [i for i in x if i >= 250]
fig.add_scatter(x=x_ekor, y=[float(density(X)(i)) for i in x_ekor], fill="tozeroy", mode="none",
                fillcolor="rgba(211,47,47,0.4)", name="P(X>250)")
atur(fig, "Normal(200, 30)", "Waktu respons (ms)", "Kepadatan")
```

```python
rata = 5
X = Exponential("X", Rational(1, 5))   # laju = 1/rata-rata

print(f"P(X>8)  = {float(P(X > 8)):.4f}  (rumus e^(-8/5) = {math.exp(-8/5):.4f})")
print(f"E(X)    = {float(E(X)):.2f}, Var(X) = {float(variance(X)):.2f}")

sampel = [random.expovariate(1 / rata) for _ in range(100_000)]
print(f"Simulasi: P(X>8) = {sum(1 for s in sampel if s > 8)/len(sampel):.4f}")

x = [i / 2 for i in range(0, 51)]
fig = go.Figure()
fig.add_histogram(x=sampel, histnorm="probability density", nbinsx=50, name="Simulasi",
                  marker_color="gray", opacity=0.5)
fig.add_scatter(x=x, y=[float(density(X)(i)) for i in x], mode="lines", name="Teori",
                line=dict(color="#d32f2f"))
fig.update_xaxes(range=[0, 25])
atur(fig, "Eksponensial(rata-rata 5 menit)", "Waktu tunggu (menit)", "Kepadatan")
```

# Bagian 4. Kesimpulan Umum

| Konsep | Pelajaran utama |
|---|---|
| Peluang dasar & kombinasi | Hitung kejadian yang diinginkan dibagi semua kemungkinan |
| Peluang bersyarat & Bayes | Informasi baru mengubah peluang |
| Kejadian independen | Peluang dikalikan |
| Simulasi Monte Carlo | Mendekati hasil teori dengan percobaan acak berulang |
| Distribusi diskrit | Binomial, Poisson, Geometrik: menghitung jumlah/kejadian |
| Distribusi kontinu | Uniform, Normal, Eksponensial: mengukur waktu/nilai berkelanjutan |

Hasil simulasi di setiap soal mendekati hasil perhitungan teori, sehingga Python terbukti bisa dipakai untuk memeriksa perhitungan manual.
